Accessing Dictionary Items
A value in a dictionary is accessed by using its key. This can be done either with square brackets [ ] or with the get() method. Both return the value linked to the given key

```python
dic = {"name":"Ashish","Age":26}
print(dic)
```

```python
dic["name"]
```

```python

dic.get("Age")
```

```python
dic1 = {"name":"Ashish","Age":26,"name":"Aashish","Age":27,"City":"Delhi","Country":"India"}
for key,value in dic1.items():
    print(key,":",value)
```

```python
d = {
    "student": {
        "name": "Sam",
        "age": 20
    },
    "Employee": {
        "name": "John",
        "age": 30
    }
}

print(d["student"]["name"], d["student"]["age"])
print(d["Employee"]["name"], d["Employee"]["age"])
```

```python
d = {
    "student": {
        "name": "Sam",
        "age": 20
    },
    "Employee": {
        "name": "John",
        "age": 30
    }
}

for key,value in d.items():
        print(key,":",value)
```

```python
d = {
    "student": {
        "name": "Sam",
        "age": 20
    },
    "Employee": {
        "name": "John",
        "age": 30
    }
}

for role, details in d.items():
    print(f"\n{role}:")
    for key, value in details.items():
        print(f"  {key}: {value}")
   
    
    
```

```python
dic = {x: x**2 for x in range(1,6)}
print(dic)
```

```python
dic = {x: x**2 for x in range(1,10) if x%2 == 0}
print(dic)
```

```python
dic = [1,1,2,2,2,3,3,3,4,4,4,5,5]
fre = {}
for i in dic:
    if i in fre:
        fre[i]+=1
    else:
        fre[i]=1
print(fre)
```

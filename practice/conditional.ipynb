natural number

```python
n=10
sum=0
count=1
while count<=n:
 sum+=count
 count+=1
print(f"sum={sum}")
```

```python
n=10
sum=0
for i in range(1,n+1):
 sum+=i
print(f"sum of natural numbers from 1 to {n} is {sum}")
```

```python
prime_count = 0

for value in range(2, 101):
    for i in range(2, value):
        if value % i == 0:
            break
    else:
        print(value)
        prime_count += 1

print(f"count = {prime_count}")
```

```python

prime_count = 0

for num in range(1,101):
    if num>1:
        for i in range(2,num):
            if (num % i) == 0:
                break
        else:
            print(num)
            prime_count += 1

print(f"count = {prime_count}")
```

```python
#list comprehension: with if condition
```

```python
lst = []
for i in range(1,11):
 if i%2==0:
  lst.append(i)
print(lst)
```

```python
even_numbers = [x for x in range(1,11) if x%2==0]
print(even_numbers)
```

# nested list comprehension

```python
list1 = [1, 2, 3, 4, 5]
list2 = ['a', 'b', 'c', 'd', 'e']
combined = [(x, y) for x in list1 for y in list2]
print(combined)
```

#list comprehension with function

```python
lst = ['apple', 'banana', 'cherry', 'date', 'elderberry']
lenghs = [len(fruit) for fruit in lst]
print(lenghs)
```

```python
nums = [10, 20, 30]

nums.append(40)      # [10,20,30,40]
print(nums)
nums.insert(1, 15)   # [10,15,20,30,40]
print(nums)
nums.remove(20)      # Removes 
print(nums)
nums.pop()           # Removes last element
print(nums)
nums.sort()          # Sort ascending
print(nums)
nums.reverse()       # Reverse order
print(nums)
nums.count(10)       # Count occurrences
print(nums.count(10))
nums.index(30)       # Find index of 30
print(nums.index(30))
```

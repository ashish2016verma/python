#there are two types of argument that is positional arguments and keyword arguments

```python
#positional argument
def positional_argument(*args):
    for i in args:
        print(i)
```

```python
positional_argument(1,2,3,4,5,"Ashish")
```

#keywords arguments

```python
def keyword_argument(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)
```

```python
keyword_argument(name="Ashish", age=25, city="Delhi")
```

```python
def both_argument(*args, **kwargs):
    for i in args:
        print(f"positional argument: {i}")
    for key, value in kwargs.items():
        print(f"keyword argument: {key} = {value}")
```

```python
both_argument(1,2,3, name="Ashish", age=25, city="Delhi")
```

```python
def temp_convert(value, unit):
    if unit.upper() == "C":
        return (value * 9/5) + 32
    elif unit.upper() == "F":
        return round((value - 32) * 5/9, 2)
    else:
        return "Invalid unit"
```

```python
value = int(input("Enter the temperature value: "))
unit = input("Enter the unit (C/F): ")
result = temp_convert(value, unit)
if result != "Invalid unit":
    print(f"The converted temperature is: {result}")
```

```python
def check_strong_password(password):
    if len(password) < 8:
        return "Password must be at least 8 characters long."
    if not any(char.isupper() for char in password):
        return "Password must contain at least one uppercase letter."
    if not any(char.islower() for char in password):
        return "Password must contain at least one lowercase letter."
    if not any(char.isdigit() for char in password):
        return "Password must contain at least one digit."
    if not any(char in "!@#$%^&*()-_=+[]{}|;:'\",.<>?/" for char in password):
        return "Password must contain at least one special character."
    return "Password is strong."
```

```python
password = input("Enter a password to check: ")
result = check_strong_password(password)
print(result)
```

```python
def calculate_bill():
    """This function calculates the total bill amount based on user input for items, their prices, and quantities."""
    n = int(input("Enter the number of items: "))
    total_cost = 0
    lst = []
    for i in range(n):
        item_name = input(f"Enter the name of item {i+1}: ")
        item_price = float(input(f"Enter the price of {item_name}: "))
        item_quantity = int(input(f"Enter the quantity of {item_name}: "))
        lst.append((item_name, item_price, item_quantity))
        total_cost += item_price * item_quantity
    return lst, round(total_cost, 2)
```

```python
result = calculate_bill()

print("Itemized Bill:")
for item in result[0]:
    print(f"{item[0]}: ${item[1]} x {item[2]} = ${item[1] * item[2]}")
print(f"The total bill amount is: ${result[1]}")
gst = result[1] * 0.18
print(f"total payment including GST (18%): ${result[1] + gst}")
```

```python
def is_palindrome(s):
    s = s.lower().replace(" ","")
    return s == s[::-1]
print(is_palindrome("Ashish"))
print(is_palindrome("A man a plan a canal Panama"))
```

```python
# check factorial using recursion
def factorial(n):
    if n== 0 or n==1:
        return 1
    else:
        return n * factorial(n-1)
print(factorial(5))  # Output: 120    
```

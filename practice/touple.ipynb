```python
#PACKING AND UNPACKING OF TOUPLE
pack_touple = 1,"Hello",3.14
print(pack_touple)
```

```python
#unpacking a tuple
a,b,c = pack_touple
print(a)
print(b)
print(c)
```

```python
n = int(input("Enter your number of feedback:"))
feedback= []
for i in range(n):
    feedbacks=input(f"Enter your feedback:{i+1}").lower()
    feedback.append(feedbacks)
fre = {}  
for j in feedback:
    if j in fre:
        fre[j]+=1  
    else:
        fre[j]=1   
print("\nYour feedbacks are:")
for i in feedback:
    print(i)
print("\nFeedback counts:")
for i, count in fre.items():
    print(f"{i}: {count}")
```

```python
n = int(input("Enter the number of elements in the tuple: "))
fre = {}
for i in range(n):
    fb = input(f"Enter feedback {i+1}: ").lower()
    fre[fb] = fre.get(fb, 0) + 1
print("\nFeedback counts:")
for feedback, count in fre.items():
    print(f"{feedback}: {count}")
```
